### Домашнее задание 9: обратная кинематика

Cопроводите решения задач короткими видео с демонстрацией работы.

**Задача 0**

Поставьте симулятор Webots (он кроссплатформенный, надо скачать установочный файл и нажимать "Вперед").

Разберитесь с интерфейсом, он в целом неплохой. Для того, чтобы открыть робота, который понадобится нам в задаче 1, нужно зайти в меню File - Open Sample World - Robots - niryo - ned - ned.wbt.

.wbt - файл т.н. "мира", Worlf File. Еще один такой лежит в папке с домашкой, это платформа Стюарта с шаром. Дефолтная платформа без шара.

Параметры роботов (длины звеньев, что откуда отсчитывается и все в таком духе) надо будет искать в гугле и в документации.

**Задача 1**

Выведите обратную кинематику для руки с 2 степенями свободы, воспользовавшись теоремой косинусов. Выберите только то решение, где первый угол больше второго.

Выведите обратную кинематику для руки с 3 степенями свободы, когда добавленная третья поворачивает плоскость, в которой находится плоский двухзвенный манипулятор.

Реализуйте аналитическую обратную кинематику на трех степенях свободы манипулятора Ned в Webots. Он должен сбивать синюю доминошку. Если переставить ее в другие координаты и подать их в вашу программу, он должен сбить ее и там.

**Задача 2**

Выведите обратную кинематику для платформы Стюарта.

**Задача 3**

Реализуйте стабилизацию шара на платформе Стюарта с помощью PD-контроллера. Мир в Webots в формате .wbt лежит в репозитории. Нужно будет написать контроллер, который считывает из симулятора положение шара и наклоняет платформу в нужную сторону. Рекомендуемая последовательность шагов:
1) запрогайте обратную кинематику
2) убедитесь, что платформа дайствительно наклоняется так, как надо
3) найдите разумные значения коэффициентов обратной связи для определения угла наклона в зависимости от отклонения шара
4) сделайте движение по квадрату
